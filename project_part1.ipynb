## Overview
Our project aims to explore and predict the impact of weather conditions on bike-sharing demand using data from the Capital Bikeshare program in Washington, D.C. By combining bike usage data with historical weather data, we seek to understand how factors such as temperature, precipitation, and wind affect the number of daily and hourly bike rentals.

The project involves cleaning and merging the datasets, conducting exploratory data analysis, and applying machine learning models to forecast demand based on weather conditions. Additionally, we will investigate differences in user behavior between casual riders and registered members, and optimize bike allocation across stations based on predicted demand fluctuations. Ultimately, our goal is to provide insights that can help bike-sharing operators better manage resources, improve user experience, and enhance service efficiency.



## Motivation

We selected the Capital Bikeshare dataset combined with weather data because bike-sharing services are an increasingly vital component of urban transportation systems, providing a flexible and eco-friendly option for city dwellers and visitors alike. However, the effectiveness and user satisfaction of these services are highly dependent on various factors, particularly weather conditions. By analyzing how external variables like temperature, precipitation, and wind impact bike-sharing demand, we can develop predictive models to optimize resource management and enhance operational efficiency. Understanding the influence of weather on user behavior could help operators make data-driven decisions to improve service delivery, ultimately benefiting both users and the business.

Through this project, we aim to explore how weather conditions drive bike-sharing demand and how user segments (e.g., casual vs. registered users) react differently to environmental changes. By answering the following key real-world questions, we hope to offer valuable insights into optimizing bike-sharing services in urban areas.

### Three Real-World Questions:

1. **How can we accurately predict daily or hourly bike-sharing demand based on weather conditions?**  
   *By answering this question, we aim to develop a machine learning model that forecasts bike-sharing demand in varying weather conditions. The outcome could help operators prepare for high or low-demand days by adjusting fleet availability, ultimately improving user satisfaction and reducing operational inefficiencies.*

2. **What are the key behavioral differences between casual and registered users in response to different weather conditions?**  
   *This question explores how different types of users—casual riders versus registered members—respond to weather variations. The findings will help operators tailor marketing strategies, promotions, or service adjustments to each group, improving overall ridership and engagement.*

3. **How can we optimize bike distribution across stations by predicting station-level demand fluctuations throughout the day?**  
   *By predicting station-level demand changes based on time of day and weather, we hope to recommend strategies for optimizing bike allocation across different stations, ensuring that bikes are available where and when they are needed most.*


# Project Part I: Data description & manipulation


## Data Sources

For this project, we are using four datasets that provide complementary information about bike-sharing usage, station locations, and weather conditions:

1. **Capital Bikeshare Ride Data (May 2020 - August 2024)**  
   This dataset contains detailed records of individual bike rides, including information on ride start and end times, bike types, station information, and user type (casual or registered).  
   - **Source**: [Capital Bikeshare System Data](https://capitalbikeshare.com/system-data)  

2. **Capital Bikeshare Station List**  
   This dataset includes the list of all stations in the Capital Bikeshare system, including station IDs and names. It is organized from the ride data and provides contextual information for ride start and end locations.  
   - **Source**: [Organized from the daily rent data](https://www.kaggle.com/datasets/taweilo/capital-bikeshare-dataset-202005202408/data?select=station_list.csv) 

3. **Station Usage Frequency Data**  
   This dataset provides daily counts of bike pickups and drop-offs at each station, offering insights into the usage patterns at specific stations over time.  
   - **Source**: [Organized from the ride data](https://www.kaggle.com/datasets/taweilo/capital-bikeshare-dataset-202005202408/data?select=usage_frequency.csv)   

4. **Weather Data (May 2020 - August 2024)**  
   This dataset includes weather-related variables for Washington, D.C., such as temperature, precipitation, wind speed, and other conditions that could influence bike usage on a daily basis.  
   - **Source**: [Visual Crossing Weather Data](https://www.visualcrossing.com/)  

### How the Datasets Complement Each Other

The four datasets complement each other by combining information on bike-sharing behavior, station-level activity, and external factors like weather. The **ride data** provides granular details about individual trips, while the **station list** and **usage frequency data** give station-level context, helping to analyze patterns of bike usage at different locations. The **weather data** adds another layer of information, enabling us to investigate how environmental conditions affect overall demand and station-level activity.

By merging these datasets, we can perform detailed analyses to predict bike-sharing demand, identify trends across different stations, and optimize resource allocation based on weather conditions and usage patterns.


## Data Description


### `daily_rent_detail.csv` dataset 
It contains information on individual bike rides from the Capital Bikeshare program from **April 30, 2020** to **August 31, 2024**. The dataset includes **16.1 million records** with detailed information about each ride, including start and end times, bike types, and station data. Below are the key variables of interest:

- **ride_id**: Unique identifier for each ride.  
  *(16.1 million unique values)*
  
- **rideable_type**: Type of bike used for the ride (e.g., classic bike, electric bike).  
  - *60% classic_bike*  
  - *29% electric_bike*  
  - *12% other*

- **started_at**: Start date and time of the ride.  
  *(Range: April 30, 2020, to August 31, 2024, Mean: November 22, 2022)*

- **ended_at**: End date and time of the ride.  
  *(Range: January 1, 2000, to August 31, 2024, Mean: November 22, 2022)*

- **start_station_name**: Name of the station where the ride started.  
  *(91% valid, 9% missing values)*

- **start_station_id**: Unique identifier for the starting station.  
  *(91% valid, 9% missing values)*

- **end_station_name**: Name of the station where the ride ended.  
  *(90% valid, 10% missing values)*

- **end_station_id**: Unique identifier for the ending station.  
  *(90% valid, 10% missing values)*

- **start_lat**: Latitude of the starting station.  
  *(Valid, no missing values)*

- **start_lng**: Longitude of the starting station.  
  *(Valid, no missing values)*

This dataset provides a wealth of information about bike-sharing behavior, and some variables (like station names and IDs) include missing values that should be addressed during data cleaning. The ride details such as timestamps and geospatial data are crucial for analyzing trip durations, distances, and the impact of various factors (like weather) on bike-sharing demand.

### `station_list.csv` dataset

The `station_list.csv` dataset contains information about the stations in the Capital Bikeshare program. The dataset includes **916 records** with details about the station IDs and names. Below are the key variables of interest:

- **station_id**: Unique identifier for each station.  
  - *(Range: 30.2k to 33.2k, Mean: 31.6k)*

- **station_name**: Name of each station.  
  - *(912 unique values)*

This dataset provides information necessary to map ride data to specific stations and locations. All records are valid, with no missing values.

### `usage_frequency.csv` dataset

It provides daily pickup and dropoff counts at each station in the Capital Bikeshare system from **April 30, 2020** to **August 30, 2024**. The dataset includes **873,000 records** with information about station names, pickup, and dropoff counts. Below are the key variables of interest:

- **date**: The date for which the data is recorded.  
  *(Range: April 30, 2020, to August 30, 2024)*

- **station_name**: The name of the station where the pickups and dropoffs occurred.  
  *(908 unique station names)*

- **pickup_counts**: The daily count of pickups at the station.  
  *(Range: 1 to 444, Mean: 16.8)*

- **dropoff_counts**: The daily count of dropoffs at the station.  
  *(Range: 0 to 460, Mean: 16.6)*

This dataset helps analyze the usage patterns at specific stations over time. All records are valid with no missing or mismatched values.


### `Weather` Data
The weather dataset used in this project contains historical weather data for Washington, D.C., from **April 30, 2020** to **August 30, 2024**. The dataset includes **1,584 records** and contains no missing or mismatched values. Below are the key variables of interest:

- **datetime**: Date corresponding to the weather data.

- **tempmax**: Maximum temperature recorded at the location (°C). 
  *(Range: -6.2°C to 39.3°C, Mean: 20.9°C)*

- **tempmin**: Minimum temperature recorded at the location (°C).  
  *(Range: -12.9°C to 27.5°C, Mean: 12°C)*

- **temp**: Average daily temperature (°C).  
  *(Range: -9.7°C to 32.7°C, Mean: 16.2°C)*

- **feelslikemax**: Maximum "feels like" temperature accounting for heat index or wind chill (°C).  
  *(Range: -11.1°C to 43.4°C, Mean: 21.1°C)*

- **feelslikemin**: Minimum "feels like" temperature (°C).  
  *(Range: -20.2°C to 31°C, Mean: 10.7°C)*

- **feelslike**: Average "feels like" temperature for the day (°C).  
  *(Range: -16.4°C to 36.3°C, Mean: 15.7°C)*

- **dew**: Dew point temperature (°C).  
  *(Range: -19.2°C to 24.2°C, Mean: 9.73°C)*
  
- **humidity**: Relative humidity as a percentage.  
  *(Range: 24.3% to 96.9%, Mean: 63.8%)*
