## Overview
Our project aims to explore and predict the impact of weather conditions on bike-sharing demand using data from the Capital Bikeshare program in Washington, D.C. Ultimately, our goal is to provide insights that can help bike-sharing operators better manage resources, improve user experience, and enhance service efficiency.

## Data Description


### `daily_rent_detail.csv` dataset 
It contains information on individual bike rides from the Capital Bikeshare program from **April 30, 2020** to **August 31, 2024**. The dataset includes **16.1 million records** with detailed information about each ride, including start and end times, bike types, and station data. Below are the key variables of interest:

- **ride_id**: Unique identifier for each ride.  
  *(16.1 million unique values)*
  
- **rideable_type**: Type of bike used for the ride (e.g., classic bike, electric bike).  
  - *60% classic_bike*  
  - *29% electric_bike*  
  - *12% other*

- **started_at**: Start date and time of the ride.  
  *(Range: April 30, 2020, to August 31, 2024, Mean: November 22, 2022)*

- **ended_at**: End date and time of the ride.  
  *(Range: January 1, 2000, to August 31, 2024, Mean: November 22, 2022)*

- **start_station_name**: Name of the station where the ride started.  
  *(91% valid, 9% missing values)*

- **start_station_id**: Unique identifier for the starting station.  
  *(91% valid, 9% missing values)*

- **end_station_name**: Name of the station where the ride ended.  
  *(90% valid, 10% missing values)*

- **end_station_id**: Unique identifier for the ending station.  
  *(90% valid, 10% missing values)*

- **start_lat**: Latitude of the starting station.  
  *(Valid, no missing values)*

- **start_lng**: Longitude of the starting station.  
  *(Valid, no missing values)*

This dataset provides a wealth of information about bike-sharing behavior, and some variables (like station names and IDs) include missing values that should be addressed during data cleaning. The ride details such as timestamps and geospatial data are crucial for analyzing trip durations, distances, and the impact of various factors (like weather) on bike-sharing demand.

### `station_list.csv` dataset

The `station_list.csv` dataset contains information about the stations in the Capital Bikeshare program. The dataset includes **916 records** with details about the station IDs and names. Below are the key variables of interest:

- **station_id**: Unique identifier for each station.  
  - *(Range: 30.2k to 33.2k, Mean: 31.6k)*

- **station_name**: Name of each station.  
  - *(912 unique values)*

This dataset provides information necessary to map ride data to specific stations and locations. All records are valid, with no missing values.

### `usage_frequency.csv` dataset

It provides daily pickup and dropoff counts at each station in the Capital Bikeshare system from **April 30, 2020** to **August 30, 2024**. The dataset includes **873,000 records** with information about station names, pickup, and dropoff counts. Below are the key variables of interest:

- **date**: The date for which the data is recorded.  
  *(Range: April 30, 2020, to August 30, 2024)*

- **station_name**: The name of the station where the pickups and dropoffs occurred.  
  *(908 unique station names)*

- **pickup_counts**: The daily count of pickups at the station.  
  *(Range: 1 to 444, Mean: 16.8)*

- **dropoff_counts**: The daily count of dropoffs at the station.  
  *(Range: 0 to 460, Mean: 16.6)*

This dataset helps analyze the usage patterns at specific stations over time. All records are valid with no missing or mismatched values.


### `Weather` Data
The weather dataset used in this project contains historical weather data for Washington, D.C., from **April 30, 2020** to **August 30, 2024**. The dataset includes **1,584 records** and contains no missing or mismatched values. Below are the key variables of interest:

- **datetime**: Date corresponding to the weather data.

- **tempmax**: Maximum temperature recorded at the location (°C). 
  *(Range: -6.2°C to 39.3°C, Mean: 20.9°C)*

- **tempmin**: Minimum temperature recorded at the location (°C).  
  *(Range: -12.9°C to 27.5°C, Mean: 12°C)*

- **temp**: Average daily temperature (°C).  
  *(Range: -9.7°C to 32.7°C, Mean: 16.2°C)*

- **feelslikemax**: Maximum "feels like" temperature accounting for heat index or wind chill (°C).  
  *(Range: -11.1°C to 43.4°C, Mean: 21.1°C)*

- **feelslikemin**: Minimum "feels like" temperature (°C).  
  *(Range: -20.2°C to 31°C, Mean: 10.7°C)*

- **feelslike**: Average "feels like" temperature for the day (°C).  
  *(Range: -16.4°C to 36.3°C, Mean: 15.7°C)*

- **dew**: Dew point temperature (°C).  
  *(Range: -19.2°C to 24.2°C, Mean: 9.73°C)*
  
- **humidity**: Relative humidity as a percentage.  
  *(Range: 24.3% to 96.9%, Mean: 63.8%)*

### Preliminary data-processing

In this step, we did some basic data processing on all tables. This includes removing data from other times according to the study period, modifying the data types of some columns, etc.
We hope that after this step, the tables will be able to be merged.

- **For usage_frequency df**：We primarily intercepted data from the study time period. Since we may need to time period, we output data for a month (from 2024-05-01 to 2024-05-31) and a year (from 2023-05-01 to 2024-05-01).
- **For weather df**：We primarily intercepted data from the study time period (from 2023-05-01 to 2024-05-01). And we also remove some columns that we may not need. We have also changed the names of some of the columns for better merge.
- **For rent_data df**：We transfer the station code from float data to int data. During the checking process, we found that some pickup times contain milliseconds information, we would like to remove this useless information. We also extracted the day information from this time column and stored it in a new ‘started_day’ column for subsequent data merging.

```python
usage_frequency['date'] = pd.to_datetime(usage_frequency['date'])

usage_frequency_m = usage_frequency.loc[(usage_frequency['date'] >= datetag2) & (usage_frequency['date'] <= datetag3)]
usage_frequency_m = usage_frequency_m.rename(columns={'station_name': 'start_station_name', 'date': 'started_day'})

usage_frequency_y = usage_frequency.loc[(usage_frequency['date'] >= datetag1) & (usage_frequency['date'] <= datetag2)].copy()
usage_frequency_y.loc[:, 'city_name'] = 'Washington,DC,USA'
```

```python
weather['datetime'] = pd.to_datetime(weather['datetime'])

weather=weather.loc[(weather['datetime']>=datetag1) & (weather['datetime']<=datetag2)]
weather=weather.iloc[:,:-1]
weather.rename(columns={'name':'city_name','datetime':'date'},inplace=True)
```

```python
rent_data['started_at'] = rent_data['started_at'].str.split('.').str[0]
rent_data['ended_at'] = rent_data['ended_at'].str.split('.').str[0]

rent_data['started_at'] = pd.to_datetime(rent_data['started_at'], format="%Y-%m-%d %H:%M:%S", errors='coerce')
rent_data['ended_at'] = pd.to_datetime(rent_data['ended_at'], format="%Y-%m-%d %H:%M:%S", errors='coerce')
rent_data['started_day'] = rent_data['started_at'].dt.floor('D')


rent_data['start_station_id'] = rent_data['start_station_id'].where(rent_data['start_station_id'].notnull(), None).astype('Int64')
rent_data['end_station_id'] = rent_data['end_station_id'].where(rent_data['end_station_id'].notnull(), None).astype('Int64')
```

## Data Visualization

### Rental Activity Overview

To understand the demand cycles, we analyzed bike rentals across different time intervals: daily, weekly, and hourly. 

These visualizations highlight the busiest periods, allowing us to identify peak hours and days of the week when rentals are most frequent.

#### Weekly Trend

```python
detail_frequency['day_of_week'] = detail_frequency['started_at'].dt.day_name()
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
detail_frequency['day_of_week'] = pd.Categorical(detail_frequency['day_of_week'], categories=day_order, ordered=True)
weekly_trends = detail_frequency.groupby('day_of_week', observed=False).size()

plt.figure(figsize=(10, 6))
plt.bar(weekly_trends.index, weekly_trends, color='c')
plt.title('Weekly Trend of Rentals')
plt.xlabel('Day of Week')
plt.ylabel('Number of Rentals')
plt.grid(axis='y')
plt.show()
```

The chart shows the number of bike rentals for each day of the week.The chart shows that bike rentals are highest during the weekdays, peaking on Thursday. Weekend rentals, especially on Saturday, drop significantly.

This may suggest that the bike-sharing system is mainly used for weekday commuting, with much less activity on weekends, likely indicating lower use for leisure or recreational purposes.

#### Daily Trend

```python
detail_frequency['date'] = detail_frequency['started_at'].dt.date
daily_rentals = detail_frequency.groupby('date').size()
plt.figure(figsize=(10, 5))
plt.plot(daily_rentals.index, daily_rentals.values, marker='o', linestyle='-',color='c')
plt.title('Daily Bike Rentals Over Time')
plt.xlabel('Date')
plt.ylabel('Number of Rentals')
plt.grid(True)
plt.show()
```

Daily rentals fluctuate sharply throughout the month. Peaks may suggest high commuting or leisure use, while dips indicate lower activity on weekends or bad weather.

#### Hourly Trend

```python
detail_frequency['hour'] = detail_frequency['started_at'].dt.hour
hourly_rentals = detail_frequency.groupby([detail_frequency['started_at'].dt.date, 'hour']).size().unstack(fill_value=0)
average_hourly_rentals = detail_frequency.groupby('hour').size()

plt.figure(figsize=(10, 5))
average_hourly_rentals.plot(kind='bar', color='c')
plt.title('Average Hourly Bike Rentals')
plt.xlabel('Hour of the Day')
plt.ylabel('Average Number of Rentals')
plt.xticks(rotation=0)
plt.show()
```

There are two peaks in the day—around 8 AM and a higher peak between 5-6 PM, aligning with typical commute times. Rentals drop after 6 PM, with the lowest activity between 1-5 AM. The bike-sharing system is mainly used for commuting, with high demand during morning and evening rush hours on workdays.

### Rental Durations


```python
detail_frequency["duration"] = (
    detail_frequency["ended_at"] - detail_frequency["started_at"]
).dt.total_seconds() / 60
detail_frequency["is_overnight"] = (
    detail_frequency["ended_at"].dt.date > detail_frequency["started_at"].dt.date
)
overnight_count = detail_frequency["is_overnight"].sum()
print(f"Number of overnight rentals: {overnight_count}")
```

```python
same_day_rentals = detail_frequency[~detail_frequency["is_overnight"]]

plt.figure(figsize=(10, 5))
sns.histplot(data=same_day_rentals, x="duration", bins=50, color='c')
plt.title("Distribution of Same-Day Rental Durations (Excluding Overnight)")
plt.xlabel("Duration (minutes)")
plt.ylabel("Frequency")
plt.show()
```

After removing overnight rentals, the majority of bike rentals are short, indicating typical usage for quick trips. There are still a few outliers with longer durations, but these are rare.

### Rental Duration by User Type

```python
detail_frequency['log_duration'] = np.log1p(detail_frequency['duration'])

plt.figure(figsize=(10, 6))
sns.boxplot(data=detail_frequency, x="member_casual", y="log_duration", color='c')
plt.title("Rental Duration by User Type (Log-Transformed)")
plt.xlabel("User Type")
plt.ylabel("Log of Duration (minutes)")
plt.show()
```

We use a log transformation to handle the wide range of rental durations and reduce the effect of extreme values. Casual users have longer and more variable rental durations than members, who tend to take shorter, more consistent trips. Casual users are likely using the service for leisure, while members use it primarily for commuting or regular short rides.

### Weather Conditions Overview

We created preliminary scatter plots to analyze the relationship between precipitation, average temperature, and pickup counts. This helps us understand whether environmental factors influence bike-sharing activity.

```python
weather_frequency['temp_avg'] = (weather_frequency['tempmax'] + weather_frequency['tempmin']) / 2
plt.figure(figsize=(10, 6))
sns.scatterplot(x='temp_avg', y='pickup_counts', data=weather_frequency, color='c')
plt.title('Average Temperature vs. Pickup Counts')
plt.xlabel('Average Temperature (°C)')
plt.ylabel('Pickup Counts')
plt.show()
```

As temperature increases, bike pickups rise, with the most activity between 10°C and 25°C, though higher temperatures still show variability. Moderate to warm temperatures lead to more bike rentals, but extreme temperatures see less activity, and temperature alone doesn't account for all variations in rentals.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='precip', y='pickup_counts', data=weather_frequency, alpha=0.6, color='c')
plt.title('Precipitation vs. Pickup Counts')
plt.xlabel('Precipitation (mm)')
plt.ylabel('Pickup Counts')
plt.show()
```

The plot shows that bike rentals significantly decrease as precipitation increases. Most rentals occur when there is little to no rain (0-10mm). After about 20 mm of precipitation, the number of pickups drops sharply, and very few rentals are seen beyond 30 mm. This suggests that rain has a clear negative impact on bike-sharing usage, with people avoiding rentals on rainy days
